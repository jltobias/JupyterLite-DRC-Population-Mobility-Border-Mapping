```python
import numpy as np
import matplotlib.pyplot as plt

# Compare population captured by successive facility buffers; use population-grid data for a real catchment calculation.
radii_km = np.array([5, 10, 25, 50])
population_captured = np.array([18000, 42000, 98000, 165000])
plt.plot(radii_km, population_captured, marker='o', color='#147d7e')
plt.fill_between(radii_km, population_captured, alpha=0.15, color='#147d7e')
plt.title('Illustrative facility catchment sensitivity')
plt.xlabel('Buffer radius (km)')
plt.ylabel('Population within catchment')
plt.grid(alpha=0.25)
plt.show()
```

## HDX Ebola cases and deaths
The following cells load the complete HDX/INRB-UMIE CSV for dated case/death summaries. The mapped layer is a locality subset because the source CSV does not include coordinates; do not interpret these summaries as risk estimates or infer transmission from mobility.

```python
import pandas as pd

ebola_url = 'https://raw.githubusercontent.com/jltobias/JupyterLite-DRC-Population-Mobility-Border-Mapping/main/data/ebola/drc_ebola_cases_consolidated.csv'
ebola = pd.read_csv(ebola_url, parse_dates=['reference_date'])
ebola = ebola[ebola['measure'].isin(['cases', 'deaths'])].copy()
ebola[['location_name', 'reference_date', 'measure', 'case_classification', 'time_period', 'value']].head()
```

```python
import matplotlib.pyplot as plt

weekly = ebola.groupby(['reference_date', 'measure'], as_index=False)['value'].sum().pivot(index='reference_date', columns='measure', values='value').fillna(0)
weekly.plot(figsize=(10, 4), color={'cases': '#d73027', 'deaths': '#542788'}, title='HDX DRC Ebola cases and deaths over time')
plt.ylabel('Reported cumulative value')
plt.xlabel('Reference date')
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

latest_date = ebola['reference_date'].max()
latest = ebola[ebola['reference_date'].eq(latest_date)].groupby(['location_name', 'measure'], dropna=False)['value'].sum().unstack(fill_value=0)
latest.sort_values('cases', ascending=True).tail(15).plot.barh(y=['cases', 'deaths'], figsize=(9, 6), color=['#d73027', '#542788'], title=f'Latest HDX Ebola values by locality ({latest_date.date()})')
plt.xlabel('Reported cumulative value')
plt.tight_layout()
plt.show()
```
