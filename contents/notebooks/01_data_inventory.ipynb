# Data inventory and provenance
This notebook records the sources needed for DRC cross-border mobility analysis. It runs in JupyterLite and does not silently claim that a live Ebola event exists. Verify event status with WHO and national authorities before analysis.

```python
from datetime import date
import pandas as pd

sources = pd.DataFrame([
    {'provider': 'IOM DTM', 'use': 'mobility and displacement observations', 'url': 'https://dtm.iom.int/'},
    {'provider': 'WorldPop', 'use': 'gridded population exposure', 'url': 'https://www.worldpop.org/'},
    {'provider': 'HDX/OCHA', 'use': 'humanitarian and boundary layers', 'url': 'https://data.humdata.org/'},
    {'provider': 'UNHCR', 'use': 'refugee and displacement context', 'url': 'https://data.unhcr.org/'},
    {'provider': 'WHO', 'use': 'outbreak verification and health context', 'url': 'https://www.who.int/emergencies/disease-outbreak-news'},
    {'provider': 'OpenStreetMap', 'use': 'roads, facilities, airports and bus stations', 'url': 'https://www.openstreetmap.org/'},
    {'provider': 'healthsites.io', 'use': 'health facility locations', 'url': 'https://healthsites.io/api/docs/'}
])
sources['checked_on'] = date.today().isoformat()
sources
```

## Minimum provenance fields
For every downloaded extract, preserve provider, title, URL, retrieval date, reference period, geography, license, and transformation steps. Aggregate mobility data cannot establish individual travel, nationality, or infection status.
