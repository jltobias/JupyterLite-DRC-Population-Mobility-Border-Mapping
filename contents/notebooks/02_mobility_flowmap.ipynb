# DRC cross-border mobility flowmap
The included values are demonstration data for checking the visualization. Replace them with a dated IOM DTM or other documented extract before interpretation.

```python
import pandas as pd

flows = pd.DataFrame([
    ('DRC', 'Uganda', 'North Kivu', 12000),
    ('DRC', 'South Sudan', 'Ituri', 2600),
    ('DRC', 'Rwanda', 'North Kivu', 8500),
    ('DRC', 'Burundi', 'South Kivu', 6200),
    ('DRC', 'Tanzania', 'Tanganyika', 4100),
    ('DRC', 'Zambia', 'Haut-Katanga', 2800),
    ('Uganda', 'DRC', 'Ituri', 3500),
], columns=['origin', 'destination', 'province_or_area', 'count'])
flows['measurement_type'] = 'demonstration only'
flows
```

```python
import matplotlib.pyplot as plt

flows.groupby('destination')['count'].sum().sort_values().plot.barh(color='#d96c42')
plt.xlabel('Illustrative count; not an official estimate')
plt.title('Cross-border mobility destinations from the demonstration table')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle

# Demonstration buffer/catchment map in lon/lat coordinates; replace points with a dated facility extract.
facilities = {'Bunia': (30.22, 1.56), 'Aru': (30.82, 2.62), 'Mahagi': (30.95, 2.28)}
fig, ax = plt.subplots(figsize=(8, 5))
for name, (lon, lat) in facilities.items():
    ax.scatter(lon, lat, s=45, label=name)
    ax.add_patch(Circle((lon, lat), 0.18, fill=False, alpha=0.45))
ax.set(title='Health-facility buffers and population-catchment review', xlabel='Longitude', ylabel='Latitude')
ax.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Choropleth-ready administrative summary; join these values to GADM geometry by ADM_ID in a GIS workflow.
admin_summary = pd.DataFrame({'admin': ['Ituri', 'North Kivu', 'South Kivu', 'Upper Nile'], 'mobility_index': [0.82, 0.67, 0.54, 0.39]})
admin_summary.sort_values('mobility_index').plot.barh(x='admin', y='mobility_index', color=plt.cm.viridis(admin_summary.sort_values('mobility_index')['mobility_index']))
plt.title('Choropleth-ready mobility index by administrative area')
plt.xlabel('Standardized demonstration index')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Compare population captured by successive facility buffers; use population-grid data for a real catchment calculation.
radii_km = np.array([5, 10, 25, 50])
population_captured = np.array([18000, 42000, 98000, 165000])
plt.plot(radii_km, population_captured, marker='o', color='#147d7e')
plt.fill_between(radii_km, population_captured, alpha=0.15, color='#147d7e')
plt.title('Illustrative facility catchment sensitivity')
plt.xlabel('Buffer radius (km)')
plt.ylabel('Population within catchment')
plt.grid(alpha=0.25)
plt.show()
```
